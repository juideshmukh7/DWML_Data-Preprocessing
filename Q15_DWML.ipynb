```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv("bookstore.csv")
```

Group by transactions

```python
transaction = df.groupby("Transaction_ID")["Book_Name"].apply(list).tolist()

print(transaction)
```

```python
te = TransactionEncoder()

encoded = te.fit_transform(transaction)

basket = pd.DataFrame(encoded, columns=te.columns_)
basket.head()
```

Apriori

```python
frequent_itemset = apriori(
    basket,
    min_support = 0.4,
    use_colnames = True
)

print(frequent_itemset)
```

Association Rule

```python
rules = association_rules(
    frequent_itemset,
    metric="confidence",
    min_threshold=0.6
)
'''
print(rules[[
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"
]])
'''
rules[[
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"
]]
```

Sort by strongest rule

```python
rules = rules.sort_values(by="confidence", ascending=False)

rules[["antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"]]
```
